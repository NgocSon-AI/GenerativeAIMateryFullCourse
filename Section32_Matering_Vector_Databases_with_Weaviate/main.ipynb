<h1>🧠 LÝ THUYẾT VỀ WEAVIATE</h1>

<h3>✅ 1. Weaviate là gì?</h3>

Weaviate là một Vector Database mã nguồn mở, được thiết kế để lưu trữ và truy vấn các vector embedding hiệu quả, đặc biệt cho các ứng dụng:

+ Chatbot có trí nhớ dài hạn

+ Semantic search (tìm kiếm theo ý nghĩa)

+ Recommendation (gợi ý tương tự)

+ RAG (Retrieval-Augmented Generation)

<h3>🔎 2. Weaviate hoạt động như thế nào?</h3>

Weaviate lưu trữ:

+ Dữ liệu gốc (text, metadata,…)

+ Vector embedding tương ứng với dữ liệu

+ Cho phép bạn truy vấn vector tương đồng (similarity search) để tìm các bản ghi gần nhất

<h3>⚙️ 3. Thành phần chính trong Weaviate<h3>

| Thành phần       | Mô tả                                                             |
| ---------------- | ----------------------------------------------------------------- |
| ✅ **Object**     | Một bản ghi (gồm vector, dữ liệu gốc và metadata)                 |
| ✅ **Class**      | Giống như bảng trong SQL, định nghĩa cấu trúc dữ liệu             |
| ✅ **Property**   | Trường (field) trong class, có thể là string, number, array, v.v. |
| ✅ **Vectorizer** | Cơ chế nhúng tự động từ text → vector (OpenAI, Cohere, v.v.)      |
| ✅ **Module**     | Plugin mở rộng chức năng (vectorizer, RAG, hybrid search, …)      |


<h3>🧰 4. Vectorizer trong Weaviate</h3>

Weaviate có thể:

+ ✅ Tự động embed dữ liệu khi bạn insert

+ ✅ Hoặc bạn embed trước rồi upload vector thủ công

| Vectorizer        | Mô hình                         |
| ----------------- | ------------------------------- |
| `text2vec-openai` | OpenAI (ada-002)                |
| `text2vec-cohere` | Cohere API                      |
| `none`            | Bạn tự nhúng (custom embedding) |


<h5>🧠 5. Các loại truy vấn trong Weaviate</h5>

+ NearVector: Tìm các object gần với một vector chỉ định

+ Hybrid Search: Kết hợp giữa vector search và từ khóa

+ Filters: Truy vấn có điều kiện (theo metadata, thời gian, tên trường...)

<h3>3. Khởi tạo client</h3>

```python
import weaviate
```

```python
client = weaviate.Client(
    url="https://your-weaviate-endpoint",
    auth_client_secret=weaviate.auth.AuthApiKey("your-api-key")
)
```

<h3>4. Khai báo schema/Create Class</h3>

```python
schema = {
    "class": "Article",
    "description": "Tài liệu nội bộ",
    "vectorizer": "text2vec-openai",
    "properties": [
        {"name": "title", "dataType": ["text"]},
        {"name": "content", "dataType": ["text"]},
        {"name": "source", "dataType": ["string"]}
    ]
}

client.schema.create_class(schema)
```

<h3>5. Thêm dữ liệu (tự nhúng)</h3>

```python
data = {
    "title": "Chính sách bảo hành",
    "content": "Sản phẩm được bảo hành 12 tháng...",
    "source": "tài liệu bảo hành.pdf"
}

client.data_object.create(
    data_object=data,
    class_name="Article"
)
```

<h3>6. Truy vấn dữ liệu</h3>

```python
query_vector = [0.1, 0.2, 0.3, ..., 0.99] # VD Embedding
result = client.query.get("Article", ["title", "content"]).with_near_vector({"vector":query_vector, "certainty: 0.7"}).with_limit(3).do()

print(result)
```

<h3>7. Truy vấn hybrid (kết hợp từ khóa + vector)</h3>

```python
result = client.query.get(
    "Article", ["title", "content"]
).with_hybird(query="bảo hành thiết bị", alpha=0.7).with_limit(3).do() 
```

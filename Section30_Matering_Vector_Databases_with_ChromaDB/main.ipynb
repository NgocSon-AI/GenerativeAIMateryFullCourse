<h1>📘 Vector Database với ChromaDB — Lý thuyết & Ứng dụng</h1>

<h3>✅ I. ChromaDB là gì?</h3>

+ ChromaDB (hoặc gọi tắt là Chroma) là một vector database mã nguồn mở được thiết kế để đơn giản hóa việc lưu trữ, tìm kiếm, và truy xuất các vector nhúng (embedding vectors) — rất phù hợp cho các hệ thống sử dụng LLM, Chatbot, RAG, Semantic Search, v.v.

+ Chroma được phát triển bởi Chroma.dev, và thường được tích hợp trực tiếp với LangChain như một bộ nhớ hoặc kho tri thức.

<h3>✅ II. Mục đích của ChromaDB</h3>

+ Lưu trữ các embedding vectors kèm theo metadata

+ Cho phép tìm kiếm ngữ nghĩa (semantic search)

+ Hỗ trợ bộ nhớ dài hạn cho LLM hoặc AI Agent

+ Dễ tích hợp vào các ứng dụng AI, đặc biệt là chatbot và hệ thống RAG

<h3>✅ III. Cách hoạt động của ChromaDB</h3>

1. Lưu trữ tri thức
+ Tài liệu (PDF, text, web...) → chia nhỏ thành đoạn

+ Mỗi đoạn → chuyển thành vector nhúng bằng model embedding (vd: text-embedding-ada-002)

+ Lưu vector + metadata vào collection trong ChromaDB

2. Tìm kiếm khi truy vấn
+ Người dùng nhập câu hỏi → chuyển thành vector

+ So sánh vector câu hỏi với vector trong DB (theo cosine similarity)

+ Trả về các đoạn văn bản tương tự nhất (semantic match)

+ Kết quả thường được gửi vào LLM (như GPT) để tạo câu trả lời có ngữ cảnh

<h3>✅ IV. Cấu trúc chính của ChromaDB</h3>

+ Collection: Tập hợp các vector embedding, giống như bảng trong SQL

+ Documents: Dữ liệu gốc (vd: đoạn văn bản)

+ Embeddings: Vector biểu diễn ngữ nghĩa của documents

+ Metadata: Thông tin kèm theo (vd: nguồn, ID, chủ đề...)

```python
import chromadb
from chromadb.config import Settings

client = chromadb.Client(Settings())
collection = client.create_collection(name="my_knowledge")

collection.add(
    documents=["Paracetamol is used to reduce fever."],
    embeddings=[[...]],  # vector 768 chiều chẳng hạn
    metadatas=[{"source": "medical_doc"}],
    ids=["doc1"]
)
```

<h3>✅ V. Ưu điểm của ChromaDB</h3>   

| Ưu điểm chính                       | Mô tả                                        |

| ✅ **Linh hoạt metadata**            | Lưu trữ và lọc theo thuộc tính tùy ý         |

<h3>✅ VII. Ứng dụng thực tế của ChromaDB</h3>

| Ứng dụng                                 | Mô tả                                                               |

| **Chatbot nội bộ**                       | Lưu tri thức của doanh nghiệp và truy xuất theo truy vấn người dùng |

<h3>2. Khởi tạo client<h3>

```python
import chromadb
from chromadb.config import Settings

client = chromadb.Client(Settings())
```

<h3>3. Tạo một collection<h3>

```python
collection = client.get_or_create_collection(name="My_Documents")
```

<h3>4. Chèn dữ liệu vào ChromaDB<h3>

```python
collection.add(
    documents = [
        "Nguyen Ngoc Son is an AI Engineer.",
        "He is a person very very good",
        "Thuy Tien is his girl friends"
    ],
    metadatas = [
        {"source":"wiki"},
        {"source":"blog"},
        {"source":"docs"}   
    ],
    ids=["doc1", "doc2", "doc3"]
)
```

<h3>5. Truy vấn tìm kiếm (semantic search)<h3>

```python
result = collection.query(
    query_texts = ["Who is Ngoc Son?"],
    n_results=2
)

print(result['documents'])
print(result['distances'])
```

<h3>6. Cập nhật hoặc xóa dữ liệu<h3>

```python
collection.update(
    ids=["doc1"],
    documents=["Updated content about ChatGPT."],
    metadatas=[{"source":"updated-wiki"}]
)
```

```python
collection.delete(ids=["doc2"])
```

<h3>7. Lưu trữ vĩnh viễn (persistent mode)</h3>

```python
# Chroma mặc định dùng memory. Để lưu trên đĩa:
client = chromadb.Client(Settings(
    chromadb_db_impl="duckdb+parquet",
    persist_directory="./chroma_data"
))
client.persist()    # Luu vao o dia
```

<h3>8. Tùy chỉnh mô hình Embedding</h3>

```python
from sentence_transformers import SentenceTransformer
from chromadb.utils.embedding_functions import SentenceTransformerEmbeddingFunction

embedding_fn = SentenceTransformerEmbeddingFunction(model_name="all-MiniLM-L6-v2")

client = chromadb.Client(Settings())
collection = client.get_or_create_collection(
    name="custom_embeds",
    embedding_function=embedding_fn
)
```

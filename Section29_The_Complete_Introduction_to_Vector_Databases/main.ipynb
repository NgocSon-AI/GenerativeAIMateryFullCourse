<h1>Vector Databases</h1>

<h3>I. Vector Database là gì?</h3>

Vector database (cơ sở dữ liệu véc-tơ) là một loại cơ sở dữ liệu chuyên biệt dùng để lưu trữ, tìm kiếm và truy vấn các vector nhúng (embedding vectors) — tức là các biểu diễn dạng số của dữ liệu phi cấu trúc như văn bản, hình ảnh, âm thanh...

🔍 Thay vì lưu trữ dữ liệu dưới dạng hàng/cột như cơ sở dữ liệu truyền thống (SQL), vector DB lưu các vector nhiều chiều đại diện cho ý nghĩa của dữ liệu.

<h3>II. Tại sao cần Vector Database?</h3>

Trong AI hiện đại, đặc biệt là LLM và search thông minh:

+ Dữ liệu như văn bản, hình ảnh… thường được biến thành vector nhúng (embedding) bằng các mô hình học sâu (ví dụ: OpenAI Embedding, BERT, CLIP…).

+ Để tìm kiếm thông tin tương tự hoặc liên quan, ta cần so sánh các vector → gọi là search theo ngữ nghĩa (semantic search).

+ Vector DB giúp tìm kiếm nhanh chóng các vector gần nhau nhất trong không gian nhiều chiều → gọi là nearest neighbor search.

<h3>III. Cách hoạt động cơ bản</h3>

1. Embedding (Mã hóa dữ liệu):

    + Một đoạn văn bản, hình ảnh, âm thanh... được đưa qua một model để chuyển thành vector nhúng.

2. Lưu trữ vector:

    + Vector nhúng được lưu trong vector DB cùng với metadata (ví dụ: ID, nội dung gốc).

3. Tìm kiếm:

    + Khi người dùng nhập câu hỏi, hệ thống cũng biến nó thành vector → so sánh với các vector trong DB để tìm những cái gần nhất theo cosine similarity, Euclidean distance, v.v.

<h3>IV. Các loại truy vấn chính trong Vector DB</h3>

| Truy vấn              | Mô tả                                                    |
| --------------------- | -------------------------------------------------------- |
| **Similarity search** | Tìm k bản ghi có vector gần nhất với truy vấn            |
| **Filtering**         | Tìm vector gần nhất nhưng có điều kiện (theo metadata)   |
| **Hybrid search**     | Kết hợp giữa keyword search (full-text) và vector search |
| **Range search**      | Tìm tất cả vector nằm trong một khoảng cách xác định     |


<h3>V. Ứng dụng của Vector Database</h3>

| Lĩnh vực                | Ứng dụng                                                                   |
| ----------------------- | -------------------------------------------------------------------------- |
| **AI/LLM**              | Tìm kiếm ngữ nghĩa (semantic search), RAG (retrieval-augmented generation) |
| **Chatbot thông minh**  | Lưu trữ và truy vấn tri thức đã embedding                                  |
| **Recommender systems** | Đề xuất dựa trên độ tương đồng vector                                      |
| **Computer vision**     | So sánh hình ảnh tương tự                                                  |
| **Audio search**        | Tìm clip âm thanh giống nhau theo nhúng                                    |

<h3>VIII. Vector Database trong hệ thống AI/LLM</h3>

Được dùng chủ yếu trong kiến trúc RAG (Retrieval-Augmented Generation):

+ Index tri thức thành vector → lưu trong Vector DB

+ Khi có câu hỏi → embedding → tìm vector gần nhất → lấy nội dung tương ứng

+ Gửi lên GPT để tổng hợp trả lời → Trả lời có ngữ cảnh chính xác hơn

✅ 3. Image / Audio / Video Search (Tìm kiếm dữ liệu phi văn bản)

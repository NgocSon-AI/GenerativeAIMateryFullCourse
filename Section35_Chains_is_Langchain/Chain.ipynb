# Theory Chains in LangChain

### 1. Chain trong LangChain là gì?

Trong LangChain, Chain (chuỗi) là một cấu trúc logic giúp kết nối nhiều thành phần lại với nhau để xây dựng một pipeline xử lý phức tạp.

⚙️ Nói cách khác: Chain cho phép bạn kết hợp các mô-đun như PromptTemplate, LLM, Memory, OutputParser… thành một hệ thống liền mạch.

### 2. Thành phần cơ bản trong một Chain

Một Chain đơn giản thường gồm 3 phần:

| Thành phần                    | Vai trò                                        |
| ----------------------------- | ---------------------------------------------- |
| **PromptTemplate**            | Tạo prompt từ dữ liệu đầu vào                  |
| **LLM**                       | Mô hình ngôn ngữ xử lý prompt                  |
| **OutputParser** *(tùy chọn)* | Phân tích kết quả đầu ra nếu cần định dạng lại |


### 3. Các loại Chain phổ biến trong LangChain

| Chain                   | Mô tả                                                           |
| ----------------------- | --------------------------------------------------------------- |
| `LLMChain`              | Chuỗi đơn giản: Prompt → LLM → Output                           |
| `SimpleSequentialChain` | Chạy các Chain tuần tự, output chain này là input chain kế tiếp |
| `SequentialChain`       | Giống trên, nhưng hỗ trợ nhiều input/output                     |
| `RouterChain`           | Dùng để định tuyến câu hỏi đến các chain khác nhau tùy nội dung |
| `ConversationChain`     | Dùng với Memory để tạo chatbot có ngữ cảnh hội thoại            |
| `TransformChain`        | Xử lý logic trung gian (ví dụ: chuyển đổi dữ liệu)              |
| `RetrievalQA`           | Kết hợp RAG: truy xuất dữ liệu từ vector DB → hỏi LLM           |


###  4. Ví dụ: LLMChain cơ bản

```python
from langchain.llms import OpenAI
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain


prompt = PromptTemplate(
    input_variables=["topic"],
    template="Hãy viết một đoạn văn giới thiệu về {topic}."
)

llm = OpenAI()
chain = LLMChain(llm=llm, prompt=prompt)

output = chain.run("Tri Tue Nhan Tao")
print(output)
```

### 5. Lợi ích của việc dùng Chain

| Lợi ích              | Mô tả                                                         |
| -------------------- | ------------------------------------------------------------- |
| ✅ Tái sử dụng        | Tạo các module riêng biệt có thể lắp ráp linh hoạt            |
| ✅ Dễ mở rộng         | Kết hợp nhiều bước phức tạp như tìm kiếm, phân tích, phản hồi |
| ✅ Quản lý trạng thái | Tích hợp với Memory để lưu giữ ngữ cảnh hội thoại             |
| ✅ Logic rõ ràng      | Dễ kiểm soát pipeline AI có nhiều bước xử lý                  |


### 6. Pipeline tổng quát sử dụng Chain

[User input] 
   ↓
[PromptTemplate]
   ↓
[LLM (OpenAI, Claude...)]
   ↓
[OutputParser hoặc Trực tiếp ra Output]


Trong trường hợp nâng cao:

[User input] 
   ↓
[RouterChain] ─────────┬─────────→ [Chain A: Trả lời luật]
                       ├─────────→ [Chain B: Trả lời du lịch]
                       └─────────→ [Chain C: Hỏi dữ liệu RAG]


### 🔐 7. Chain + Memory = Agent nền tảng hội thoại

Chain có thể kết hợp với:

+ Memory: lưu lại lịch sử hội thoại

+ Tool: dùng để gọi API, search web, tính toán

+ Agent: quyết định logic xử lý nào cần thực hiện tiếp

### 8. Tổng kết

| Mục                 | Nội dung                                                          |
| ------------------- | ----------------------------------------------------------------- |
| **Chain là gì?**    | Pipeline xử lý có nhiều bước dùng để gọi và phối hợp LLM          |
| **Dùng để làm gì?** | Chatbot, hỏi-đáp, xử lý dữ liệu, tổng hợp, RAG                    |
| **Loại phổ biến**   | `LLMChain`, `ConversationChain`, `SequentialChain`, `RouterChain` |
| **Tích hợp được**   | Prompt, LLM, Tool, Memory, Agent, Vector DB                       |

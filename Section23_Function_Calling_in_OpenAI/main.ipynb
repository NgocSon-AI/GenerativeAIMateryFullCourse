<h1>Function Calling in OpenAI</h1>

<h3>🧠 Function Calling trong OpenAI là gì?</h3>

Function Calling trong OpenAI là cơ chế cho phép mô hình ChatGPT hiểu và gọi các hàm (API hoặc logic định nghĩa sẵn) từ phía hệ thống backend của bạn — một cách có kiểm soát và có cấu trúc.

Nói cách khác, bạn mô tả hàm của mình (tên, đối số, kiểu dữ liệu), và mô hình có thể quyết định khi nào và với tham số nào cần gọi hàm đó, thay vì chỉ trả về văn bản thông thường.

<h3>✅ Tại sao cần Function Calling?</h3>
Giúp mô hình tích hợp với hệ thống thực tế (truy vấn database, API thời tiết, thông tin chuyến bay, v.v.)

Đảm bảo kết quả đúng và có thể kiểm soát được

Cho phép tạo các chatbot thực thi hành động chứ không chỉ trả lời tĩnh

<h3>🔧 Cách hoạt động Function Calling</h3>

<h5>1. Bạn định nghĩa hàm bằng JSON Schema:</h5>

```python
{
  "name": "get_weather",
  "description": "Lấy thông tin thời tiết hiện tại theo vị trí",
  "parameters": {
    "type": "object",
    "properties": {
      "location": {
        "type": "string",
        "description": "Tên thành phố hoặc khu vực"
      }
    },
    "required": ["location"]
  }
}
```

<h5>2. Bạn truyền function này vào API ChatGPT:</h5>

```python
response = client.chat.completions.create(
  model="gpt-4",
  messages=[...],
  functions=[your_function_definition],
  function_call="auto"  # để mô hình tự quyết định khi nào gọi hàm
)
```

<h3>3. Mô hình trả về yêu cầu gọi hàm:</h3>

```python
{
  "role": "assistant",
  "function_call": {
    "name": "get_weather",
    "arguments": "{ \"location\": \"Hanoi\" }"
  }
}
```

<h5>4. Bạn thực thi hàm thực tế trong backend, rồi truyền kết quả trở lại:</h5>

```python
# Thực thi gọi API thời tiết thực tế
weather_data = call_real_weather_api("Hanoi")

# Truyền kết quả vào đoạn chat tiếp theo
messages.append({
    "role": "function",
    "name": "get_weather",
    "content": json.dumps(weather_data)
})

```

<h3>🧩 Ưu điểm của Function Calling</h3>

| Ưu điểm           | Mô tả                                  |
| ----------------- | -------------------------------------- |
| 🔐 An toàn hơn    | Mô hình không tự sinh API call sai     |
| 🧠 Thông minh hơn | Mô hình hiểu khi nào cần dùng function |

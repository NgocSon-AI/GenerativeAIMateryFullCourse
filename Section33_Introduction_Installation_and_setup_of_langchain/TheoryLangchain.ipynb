# 🧠 I. LÝ THUYẾT VỀ LANGCHAIN

<h3>✅ 1. LangChain là gì?</h3>

LangChain là một framework mã nguồn mở giúp xây dựng ứng dụng sử dụng LLM (Large Language Models) một cách có cấu trúc, hiệu quả và mở rộng.

<h3>🎯 2. Mục tiêu chính của LangChain</h3>

Kết hợp LLM với:

+ Dữ liệu riêng (private data)

+ Công cụ bên ngoài (API, SQL, file, browser, v.v.)

+ Workflow phức tạp (multi-step reasoning, agent, memory...)

LangChain giúp bạn không chỉ "chat với GPT" mà còn xây dựng hệ thống thông minh có khả năng ra quyết định, tìm kiếm, kết nối dữ liệu.

<h3>🧱 3. Các thành phần chính</h3>

| Thành phần         | Chức năng                                                        |
| ------------------ | ---------------------------------------------------------------- |
| **LLM**            | Giao tiếp với GPT/OpenAI/Claude/v.v.                             |
| **PromptTemplate** | Tạo prompt động theo đầu vào                                     |
| **Chain**          | Chuỗi xử lý nhiều bước: prompt → LLM → đầu ra                    |
| **Retriever**      | Lấy dữ liệu liên quan từ Vector DB                               |
| **Memory**         | Lưu và nhớ lịch sử hội thoại                                     |
| **Tool**           | Các API/công cụ mà Agent có thể dùng                             |
| **Agent**          | Thực thể thông minh dùng LLM để quyết định công việc và gọi Tool |
| **VectorStore**    | Quản lý embedding và tìm kiếm tương đồng                         |


<h3>⚙️ 4. Các kiến trúc quan trọng</h3>

🔁 a. Chain (Chuỗi)

Một pipeline bao gồm nhiều bước:

Prompt → LLM → Output Parser → Final Result

→ Thường dùng để xử lý truy vấn đơn giản hoặc custom workflow.

🧠 b. Agent
Agent dùng LLM để "tự quyết định" bước tiếp theo cần làm gì.

Ví dụ:

+ Người dùng hỏi: "Ngày mai có mưa không ở Hà Nội?"

+ Agent:

    + Nhận câu hỏi

    + Gọi tool WeatherAPI

    + Trả lời người dùng dựa vào dữ liệu gọi được

📚 c. RAG (Retrieval-Augmented Generation)
+ Kết hợp LLM với dữ liệu ngoài (ví dụ tài liệu nội bộ)

Cấu trúc:

+ Câu hỏi người dùng → Embed → Truy vấn VectorStore

+ Lấy top-k đoạn liên quan → Chèn vào prompt → Gửi tới LLM → Tạo câu trả lời

→ LangChain hỗ trợ triển khai RAG rất thuận tiện với Retriever, VectorStore, Document Loader,...

## 🔰 Cơ bản

### 1. Khởi tạo LLM

```python
from langchain.chat_models import ChatOpenAI
llm = ChatOpenAI(temperature=0)
```

### 2. Tạo PromptTemplate

```python
from langchain.prompts import PromptTemplate

template = PromptTemplate.from_template("Translate this to French: {text}")
```

### 3. Tạo Chain

```python
from langchain.chains import LLMChain
chain = LLMChain(llm=llm, prompt=template)
result = chain.run(text="Hello")
```

## ⚡ Trung cấp

### 4. RAG với VectorStore

```python
from langchain.vectorstores import FAISS
from langchain.embeddings import OpenAIEmbeddings

# Tạo embedding & lưu
db = FAISS.from_texts(["Documents 1", "Doc 2"], embedding=OpenAIEmbeddings())

# Truy vấn tương đồng
retriever = db.as_retriever()
docs = retriever.get_relevant_documents("Tìm kiếm gì đó")
```

<h3>5. Tích hợp với Chain</h3>

```python
from langchain.chains import RetrievalQA
qa_chain = RetrievalQA.from_chain_type(llm=llm, retriever=retriever)
qa_chain.run("Câu hỏi của bạn?")
```

## 🔥 Nâng cao

### 6. Dùng Agent có Tool

```python
from langchain.agents import initialize_agent, Tool
from langchain.agents.agent_types import AgentType

tools = [
    Tool(
        name="Calculator",
        func=lambda x: eval(x),
        description="Tính toán"
    )
]

agent = initialize_agent(tools, llm, agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION)
agent.run("Tính 2025 + 7")
```

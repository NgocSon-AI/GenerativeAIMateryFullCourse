# Prompt Templates In Langchain

### 🧠 1. Prompt Template trong Langchain là gì?

Prompt Template là một mẫu gợi ý có tham số (parameterized prompt) — đóng vai trò là cầu nối giữa dữ liệu đầu vào của người dùng và mô hình ngôn ngữ (LLM).

Trong Langchain, Prompt Template giúp bạn tự động hóa, linh hoạt hóa và chuẩn hóa quá trình xây dựng prompt bằng cách:

+ Cho phép xác định các biến đầu vào.

+ Dễ dàng định dạng lại thành prompt hoàn chỉnh bằng cách truyền dữ liệu cụ thể vào.

### 🧩 2. Cấu trúc của một Prompt Template

Prompt Template thường gồm 2 phần:

+ template: chuỗi văn bản có chứa biến (placeholder), ví dụ: "Tóm tắt bài viết sau: {text}"

+ input_variables: danh sách tên biến cần truyền vào (["text"])

Ví dụ:

```python
from langchain.prompts import PromptTemplate

template = "Hãy viết một đoạn giới thiệu về {topic} trong {num_words} từ."
prompt = PromptTemplate(
    input_variables=["topic", "num_words"],
    template=template
)
```

Sau đó, bạn có thể tạo prompt cụ thể bằng:

```python
prompt.format(topic="AI", num_words=100)
```

### 🧭 3. Vai trò của Prompt Template trong Langchain

| Vai trò                        | Giải thích                                                          |
| ------------------------------ | ------------------------------------------------------------------- |
| **Chuẩn hóa prompt**           | Tránh việc hardcode từng prompt, giúp quản lý tốt hơn               |
| **Tái sử dụng**                | Sử dụng cùng một template cho nhiều đầu vào khác nhau               |
| **Tối ưu hóa mô hình**         | Giúp thử nghiệm nhiều kiểu prompt nhanh chóng                       |
| **Tách biệt logic và dữ liệu** | Giúp giữ code sạch, dễ debug, dễ maintain                           |
| **Tích hợp vào Chain**         | Là thành phần cơ bản trong `LLMChain`, `RouterChain`, `Agent`, v.v. |

### 🧱 4. Các loại Prompt Template trong Langchain

| Loại Template           | Mô tả                                                                                            |
| ----------------------- | ------------------------------------------------------------------------------------------------ |
| `PromptTemplate`        | Dùng cho các LLM truyền thống (input là chuỗi văn bản)                                           |
| `ChatPromptTemplate`    | Dùng cho các mô hình chat như GPT-3.5/4, có dạng input theo role (`system`, `user`, `assistant`) |
| `FewShotPromptTemplate` | Cho phép bạn đưa các ví dụ mẫu để huấn luyện mô hình dạng few-shot                               |
| `DynamicPromptTemplate` | Tạo prompt linh hoạt theo logic runtime                                                          |
| `StringPromptTemplate`  | Base class để tạo custom prompt                                                                  |


### 📌 5. Ứng dụng thực tế

| Use Case        | Prompt Template ví dụ                                  |
| --------------- | ------------------------------------------------------ |
| Chatbot         | "Xin chào {name}, tôi có thể giúp gì cho bạn hôm nay?" |
| Summarization   | "Tóm tắt đoạn văn sau trong 3 câu: {content}"          |
| Q\&A            | "Câu hỏi: {question}\nTrả lời:"                        |
| Code generation | "Viết một hàm Python để làm điều sau: {instruction}"   |


### 🔄 6. Tích hợp với Chain

Prompt Template thường được tích hợp vào LLMChain:

```python
from langchain.chains import LLMChain
from langchain.llms import OpenAI

llm_chain = LLMChain(
    llm=OpenAI(),
    prompt=prompt
)
```

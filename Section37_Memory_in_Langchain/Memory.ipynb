# 🧠 1. Lý thuyết về Memory trong LangChain

## 1.1. Khái niệm

+ Memory là nơi lưu trữ các thông tin từ các lượt tương tác trước đó.

+ Giúp cho mô hình LLM (Large Language Model) hiểu được ngữ cảnh của cuộc trò chuyện và trả lời chính xác hơn.

+ Có thể lưu:

    + Toàn bộ lịch sử chat (ConversationBufferMemory)

    + Chỉ thông tin gần đây (ConversationBufferWindowMemory)

    + Thông tin đã được tóm tắt (ConversationSummaryMemory)

    + Kết hợp nhiều phương pháp trên

### 1.2. Một số loại Memory trong LangChain

| Loại Memory                      | Mô tả                                                                                |
| -------------------------------- | ------------------------------------------------------------------------------------ |
| `ConversationBufferMemory`       | Lưu toàn bộ lịch sử hội thoại dưới dạng văn bản.                                     |
| `ConversationBufferWindowMemory` | Lưu một **cửa sổ** gồm N lượt hội thoại gần nhất.                                    |
| `ConversationSummaryMemory`      | Tóm tắt toàn bộ hội thoại cũ thành một bản tóm gọn để mô hình hiểu nhanh.            |
| `CombinedMemory`                 | Kết hợp nhiều loại memory.                                                           |
| `VectorStoreRetrieverMemory`     | Lưu trữ thông tin dưới dạng **vector embeddings** và truy xuất bằng semantic search. |

### 2.2. Ví dụ: Dùng ConversationBufferMemory với LLMChain

```python
from langchain.chat_models import ChatOpenAI
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory


llm = ChatOpenAI(model="gpt-3.5-turbo")

memory = ConversationBufferMemory()

conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=True
)
print(conversation.predict(input="Xin chào!"))
print(conversation.predict(input="Tôi tên là Sơn."))
print(conversation.predict(input="Tên tôi là gì?"))
```

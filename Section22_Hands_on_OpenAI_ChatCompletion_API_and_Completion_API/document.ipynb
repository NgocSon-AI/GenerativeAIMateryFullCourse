<h2>✅ 1. Completion API là gì?</h2>

Completion API là API đời đầu của OpenAI, sử dụng các mô hình như text-davinci-003. Đây là giao diện đơn giản cho việc sinh văn bản theo kiểu hoàn tất đoạn văn bản đầu vào.

+ Input (prompt): là một chuỗi văn bản (text).

+ Output: mô hình sẽ “hoàn tất” đoạn văn bản đó.

🔹 Ví dụ:

```python
import openai

openai.api_key = "your-api-key"

response = openai.Completion.create(
    engine="text-davinci-003",
    prompt="Once upon a time,",
    max_tokens=50
)

print(response["choices"][0]["text"])
```

<h2>✅ 2. Chat Completion API là gì?</h2>

Chat Completion API được giới thiệu từ GPT-3.5 và GPT-4 trở đi (đặc biệt dùng với gpt-3.5-turbo, gpt-4, gpt-4o). Thay vì truyền chuỗi văn bản thẳng như Completion API, bạn gửi danh sách các tin nhắn dạng hội thoại giữa người dùng và trợ lý AI.

+ Input: danh sách các tin nhắn (messages) với vai trò system, user, assistant.

+ Output: phản hồi của mô hình dạng hội thoại.

🔹 Ví dụ:

```python
import openai

openai.api_key = "your-api-key"

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What's the weather like today?"}
    ]
)

print(response['choices'][0]['message']['content'])
```

🔸 Ưu điểm của Chat Completion API:

+ Hỗ trợ cấu trúc hội thoại nhiều lượt (multi-turn).

+ Gần gũi với trải nghiệm ChatGPT thực tế.

+ Có thể thiết lập hành vi bằng vai trò system

| Tính năng                  | **Completion API**         | **Chat Completion API**            |
| -------------------------- | -------------------------- | ---------------------------------- |
| Mô hình hỗ trợ             | `text-davinci-003`, cũ hơn | `gpt-3.5-turbo`, `gpt-4`, `gpt-4o` |
| Kiểu input                 | Chuỗi văn bản (prompt)     | Danh sách hội thoại (`messages`)   |
| Hỗ trợ hội thoại           | ❌                          | ✅                                  |
| Hiện đại, được khuyến nghị | ❌                          | ✅ (ưu tiên hiện tại)               |
